## Data Overview
- 15,420 rows, 33 columns, 0 missing values, 0 duplicates — a clean dataset
- Target: FraudFound_P (0 = no fraud, 1 = fraud)
- Fraud rate: 6.0% (923 of 15,420 claims) — imbalanced; precision/recall/PR-AUC prioritized over accuracy

### Column notes
- PolicyNumber: 15,420/15,420 unique — pure ID, excluded from modeling
- RepNumber: 16 categories, evenly distributed — kept as categorical, weak predictor expected
- Year: only 1994-1996 — old dataset; noted as a limitation
- Days_Policy_Accident: 98.9% "more than 30" — very low variance, likely weak predictor
- PastNumberOfClaims, AgeOfVehicle: ordinal categories, need custom sort order for charts
- PoliceReportFiled: 2.8% "Yes" — tests H4, but rare-category imbalance itself
- WitnessPresent: 0.6% "Yes" — extremely rare, limited statistical power expected

### Known Limitations
- Data is from 1994-1996 — a synthetic/older dataset, not representative of current fraud patterns
- FraudFound_P imbalance (6%) requires careful metric choice (precision, recall, PR-AUC, not accuracy)
- Findings show association, not proof that a factor causes fraud
