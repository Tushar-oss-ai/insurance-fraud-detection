## Data Overview
- 15,420 rows, 33 columns, 0 missing values, 0 duplicates — a clean dataset
- Target: FraudFound_P (0 = no fraud, 1 = fraud)
- Fraud rate: 6.0% (923 of 15,420 claims) — imbalanced; precision/recall/PR-AUC prioritized over accuracy

### Column notes
- PolicyNumber: 15,420/15,420 unique — pure ID, excluded from modeling
- RepNumber: 16 categories, evenly distributed — kept as categorical, weak predictor expected
- Year: only 1994-1996 — old dataset; noted as a limitation
- Days_Policy_Accident: 98.9% "more than 30" — very low variance, likely weak predictor
- PastNumberOfClaims, AgeOfVehicle: ordinal categories, need custom sort order for charts
- PoliceReportFiled: 2.8% "Yes" — tests H4, but rare-category imbalance itself
- WitnessPresent: 0.6% "Yes" — extremely rare, limited statistical power expected

### Known Limitations
- Data is from 1994-1996 — a synthetic/older dataset, not representative of current fraud patterns
- FraudFound_P imbalance (6%) requires careful metric choice (precision, recall, PR-AUC, not accuracy)
- Findings show association, not proof that a factor causes fraud

```python
print("Age == 0:", (df_clean["Age"] == 0).sum())
print(df_clean[["DayOfWeekClaimed", "MonthClaimed"]].apply(lambda s: (s == "0").sum()))
```

```python
# Age = 0 is a placeholder for "unknown", not a real age
print(pd.crosstab(df_clean["Age"] == 0, df_clean["AgeOfPolicyHolder"]))

df_clean["Age_Missing"] = (df_clean["Age"] == 0).astype(int)
df_clean["Age_Clean"] = df_clean["Age"].replace(0, np.nan)

# Drop rows with an invalid claim month/day (very few, if any)
before = len(df_clean)
df_clean = df_clean[(df_clean["MonthClaimed"] != "0") & (df_clean["DayOfWeekClaimed"] != "0")]
print("Rows dropped:", before - len(df_clean))
```

```python
ordinal_orders = {
    "AgeOfVehicle": ["new", "2 years", "3 years", "4 years", "5 years", "6 years", "7 years", "more than 7"],
    "PastNumberOfClaims": ["none", "1", "2 to 4", "more than 4"],
    "Days_Policy_Accident": ["none", "1 to 7", "8 to 15", "15 to 30", "more than 30"],
    "VehiclePrice": ["less than 20000", "20000 to 29000", "30000 to 39000", "40000 to 59000", "60000 to 69000", "more than 69000"],
    "AgeOfPolicyHolder": ["16 to 17", "18 to 25", "26 to 30", "31 to 35", "36 to 40", "41 to 50", "51 to 65", "over 65"],
    "NumberOfSuppliments": ["none", "1 to 2", "3 to 5", "more than 5"],
    "AddressChange_Claim": ["no change", "under 6 months", "1 year", "2 to 3 years", "4 to 8 years"],
    "Days_Policy_Claim": ["none", "8 to 15", "15 to 30", "more than 30"],
}

for col, order in ordinal_orders.items():
    df_clean[col + "_ord"] = pd.Categorical(df_clean[col], categories=order, ordered=True).codes
    bad = (df_clean[col + "_ord"] == -1).sum()
    print(f"{col}: unmatched = {bad}")
    if bad > 0:
        print("   actual labels:", df_clean[col].unique().tolist())
```

```python
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
m = {name: i + 1 for i, name in enumerate(months)}

df_clean["Claim_Delay_Months"] = (df_clean["MonthClaimed"].map(m) - df_clean["Month"].map(m)) % 12
df_clean["Weekend_Accident"] = df_clean["DayOfWeek"].isin(["Saturday", "Sunday"]).astype(int)
df_clean["Weekend_Claim"] = df_clean["DayOfWeekClaimed"].isin(["Saturday", "Sunday"]).astype(int)

print(df_clean["Claim_Delay_Months"].value_counts().sort_index())
```

```python
df_clean["Police_Report"] = (df_clean["PoliceReportFiled"] == "Yes").astype(int)
df_clean["Witness"] = (df_clean["WitnessPresent"] == "Yes").astype(int)
df_clean["Policyholder_At_Fault"] = (df_clean["Fault"] == "Policy Holder").astype(int)
df_clean["Agent_External"] = (df_clean["AgentType"] == "External").astype(int)

print(df_clean[["Police_Report", "Witness", "Policyholder_At_Fault", "Agent_External"]].mean().round(3))
```

```python
for col in ["PoliceReportFiled", "WitnessPresent", "Fault", "PolicyType", "BasePolicy", "AccidentArea", "AgentType"]:
    print(df_clean.groupby(col)["FraudFound_P"].agg(["count", "mean"]).round(3))
    print()
```

Every feature must be known when the claim is filed, before any investigation. If you used something learned during the investigation, the model would cheat. Everything above passes: demographics, policy details, accident details and claim timing are all known at filing. Excluded from modeling: PolicyNumber (an ID) and FraudFound_P (the target). Treat RepNumber and Year with caution: Year covers only 1994-1996, so a model could learn year-specific quirks

## Feature Dictionary

| Feature | Definition | Why it's needed |
|---|---|---|
| `Age_Clean` / `Age_Missing` | Age with impossible 0s set to missing / flag for those rows | Keeps unknown ages from being read as "age 0" |
| `*_ord` columns | Numeric rank of ordered categories (vehicle age, claim history, price, etc.) | Lets models and correlation tests use the natural order |
| `Claim_Delay_Months` | Months between accident and claim filing | Tests whether late claims are riskier |
| `Weekend_Accident`, `Weekend_Claim` | 1 if on Saturday/Sunday | Timing patterns |
| `Police_Report`, `Witness`, `Policyholder_At_Fault`, `Agent_External` | Yes/no flags | Tests H4 and other hypotheses |

**Leakage check:** all features are available at claim filing time. Excluded: PolicyNumber (ID), FraudFound_P (target).
